# Quant Research Notebook

Quick research pass on the simulated market ticks: momentum vs mean-reversion
signals and a simple backtest. Everything here runs on synthetic data -
none of this is trading advice.

```python
import sys
sys.path.append("../src")
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from quant import momentum_signals, mean_reversion_signals, summarize

# Synthetic price series so the notebook runs without Kafka.
rng = np.random.default_rng(7)
prices = 150 * np.cumprod(1 + rng.normal(0, 0.004, 2000))
df = pd.DataFrame({"price": prices})
```

```python
mom = momentum_signals(df)
mr = mean_reversion_signals(df)
summarize(mom, "momentum")
summarize(mr, "mean-reversion")
```

```python
plt.figure(figsize=(10, 4))
plt.plot(df["price"], label="price")
plt.plot(mom["sma_fast"], label="SMA 10")
plt.plot(mom["sma_slow"], label="SMA 30")
plt.legend(); plt.title("Momentum signal moving averages")
plt.show()
```

## Takeaways

- On random-walk data neither signal should show a real edge - if one does, suspect lookahead bias.
